## What is sentiment analysis?

**Sentiment analysis** is text classification for *emotion*: is the text **positive**, **negative** (or **neutral**)?

> 💬 *"I am so excited for the weekend!"*  →  😊 **POSITIVE (1.00)**

It is the most famous example in the Hugging Face ecosystem — so famous that it has its own default pipeline task:


```python
from transformers import pipeline

# No model argument needed! 'sentiment-analysis' has a canonical default:
# distilbert-base-uncased-finetuned-sst-2-english (~270 MB, cached on first use)
sentiment = pipeline("sentiment-analysis", device=-1)
```

**Analyze one text:**

```python
r = sentiment("The hotel room was spotless and the staff were wonderful.")[0]
print(f"{r['label']}  with confidence {r['score']:.3f}")
```

## Batch processing

Hand the pipeline a list of reviews and it scores them all at once:


```python
reviews = [
    "I absolutely loved the film, the acting was brilliant.",
    "The service was slow and the food was cold.",
    "The product works exactly as described, very happy.",
    "I would not recommend this to anyone.",
]
results = sentiment(reviews)
for review, r in zip(reviews, results):
    print(f"{r['label']:<9} ({r['score']:.2f})  {review!r}")
```

## Interpreting the score

The score is a **confidence**, not a truth:

- `0.99` → the model is very sure.
- `0.55` → borderline; the text is probably mixed or sarcastic.

Stress-test it — sarcasm is where sentiment models fail:


```python
tricky = [
    "Great, another update I didn't ask for.",   # sarcastic
    "The battery lasts two days. Two. Days.",     # emphatic positive
    "I waited 3 hours for this sandwich.",         # fact, negative tone
]
for t, r in zip(tricky, sentiment(tricky)):
    print(f"{r['label']:<9} ({r['score']:.2f})  {t!r}")
```

## 2-class vs. 3-class models

- **2-class (SST-2, default):** `POSITIVE` / `NEGATIVE` — forced to choose, even for neutral texts.
- **3-class:** also knows `NEUTRAL` — better for real-world data (reviews, tweets).


```python
sentiment3 = pipeline(
    "text-classification",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest",
    device=-1,
)
r = sentiment3("The train was on time for once.")
print(f"3-class model says: {r[0]['label']} ({r[0]['score']:.2f})")
```

## Try it yourself

1. Score 5 of your own texts, including one sarcastic one.
2. Compare the 2-class and 3-class predictions on the same texts.

## Common pitfalls

- **Sarcasm and irony** fool models — check borderline scores by hand.
- **Labels differ per model** — read them, don't hard-code.

## Summary

- Sentiment analysis = text classification for emotion.
- `pipeline("sentiment-analysis")` works with no model argument.
- Always look at both `label` and `score`.
